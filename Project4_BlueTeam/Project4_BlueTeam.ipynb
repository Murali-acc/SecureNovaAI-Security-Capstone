```python
#Step 22 – Create the custom input guardrail

import re
import json
import datetime
import os

from nemoguardrails.actions import action

GUARDRAIL_LOG = os.path.join(
    RESULTS_DIR,
    "guardrail_decisions.jsonl"
)

ATTACK_PATTERNS = {
    "Indirect Prompt Injection": [
        r"ignore.{0,30}(previous|prior|above|document)",
        r"reveal.{0,30}(credential|token)",
        r"include.{0,30}hidden token",
        r"print.{0,30}internal identity"
    ],

    "Agent Identity Spoofing": [
        r"FROM:\s*TRUSTED_ORCHESTRATOR",
        r"administrator mode",
        r"privileged operation",
        r"trusted orchestrator"
    ],

    "System Prompt Extraction": [
        r"system prompt",
        r"initial instructions",
        r"confidential.{0,30}instructions",
        r"repeat everything.{0,30}told"
    ],

    "RAG Poisoning": [
        r"grant.{0,30}access",
        r"poisoned.{0,30}document",
        r"follow this document.{0,30}instructions"
    ],

    "MCP Tool Abuse": [
        r"grant_demo_access",
        r"call.{0,30}tool",
        r"without additional authorization",
        r"execute.{0,30}privileged"
    ]
}

def detect_attack(text):

    for category, patterns in ATTACK_PATTERNS.items():

        for pattern in patterns:

            if re.search(pattern, text, re.IGNORECASE | re.DOTALL):

                return {
                    "blocked": True,
                    "category": category,
                    "reason": "Suspicious instruction pattern detected."
                }

    return {
        "blocked": False,
        "category": "None",
        "reason": "No configured attack pattern detected."
    }


print("Custom input guardrail rules created.")
print("Attack categories:", len(ATTACK_PATTERNS))
```

```python
#Step 29 – Implement JWT output redaction
#This control will detect strings that look like JSON Web Tokens (JWTs) in outgoing agent responses and replace them with [REDACTED].

#29A. Create the JWT detection and redaction function

import re
import datetime
import json
import os

JWT_PATTERN = (
    r"\b[A-Za-z0-9_-]{5,}\."
    r"[A-Za-z0-9_-]{5,}\."
    r"[A-Za-z0-9_-]{5,}\b"
)

def redact_jwt_tokens(text):

    redacted_text, count = re.subn(
        JWT_PATTERN,
        "[REDACTED]",
        text
    )

    return redacted_text, count

print("JWT output guardrail created successfully.")
```

#### Cryptographic agent identity binding using Ed25519

```python
#Step 30 – Generate Ed25519 key pair

import os
from pathlib import Path

from cryptography.hazmat.primitives.asymmetric.ed25519 import (
    Ed25519PrivateKey
)
from cryptography.hazmat.primitives import serialization

# Generate Ed25519 private key
private_key = Ed25519PrivateKey.generate()

# Derive public key
public_key = private_key.public_key()

# Save private key
private_key_path = os.path.join(
    KEY_DIR,
    "agent_private_key.pem"
)

with open(private_key_path, "wb") as file:
    file.write(
        private_key.private_bytes(
            encoding=serialization.Encoding.PEM,
            format=serialization.PrivateFormat.PKCS8,
            encryption_algorithm=serialization.NoEncryption()
        )
    )

# Save public key
public_key_path = os.path.join(
    KEY_DIR,
    "agent_public_key.pem"
)

with open(public_key_path, "wb") as file:
    file.write(
        public_key.public_bytes(
            encoding=serialization.Encoding.PEM,
            format=serialization.PublicFormat.SubjectPublicKeyInfo
        )
    )

print("Ed25519 key pair generated successfully!")
print("Private key file exists:", os.path.exists(private_key_path))
print("Public key file exists:", os.path.exists(public_key_path))
```

```python
# Step 31 – Sign an outgoing agent message
# 31A. Create the signing function

import base64
import datetime
import json

def sign_agent_message(message, private_key):

    message_bytes = message.encode("utf-8")

    signature = private_key.sign(message_bytes)

    signature_base64 = base64.b64encode(
        signature
    ).decode("utf-8")

    return signature_base64


agent_identity = "SecureNova_AI_Agent"

outgoing_message = (
    "SecureNova AI: Authorized agent response. "
    "Identity verification enabled."
)

signature = sign_agent_message(
    outgoing_message,
    private_key
)

print("Agent identity:", agent_identity)
print("Outgoing message:", outgoing_message)
print("Digital signature generated:", bool(signature))
print("Signature algorithm: Ed25519")
```

```python
#Step 32 – Verify the signature
#32A. Create the verification function

from cryptography.exceptions import InvalidSignature

def verify_agent_message(message, signature_base64, public_key):

    try:
        signature_bytes = base64.b64decode(
            signature_base64,
            validate=True
        )

        public_key.verify(
            signature_bytes,
            message.encode("utf-8")
        )

        return True, "Signature verification successful."

    except (InvalidSignature, ValueError):
        return False, "ERROR: Signature verification failed. Message rejected."


verified, verification_message = verify_agent_message(
    outgoing_message,
    signature,
    public_key
)

print("Verification result:", verified)
print("Message:", verification_message)
```

```python
#32B. Test message tampering
tampered_message = outgoing_message.replace(
    "Authorized",
    "Unauthorized",
    1
)

tampered_verified, tampered_result = verify_agent_message(
    tampered_message,
    signature,
    public_key
)

print("Original message:", outgoing_message)
print("Tampered message:", tampered_message)
print("Verification result:", tampered_verified)
print("Security response:", tampered_result)

if not tampered_verified:
    print("ALERT: Tampered agent message rejected!")
```

```python
#32C. Save the verification results
signature_results = {
    "timestamp": datetime.datetime.now().astimezone().isoformat(),
    "agent_identity": agent_identity,
    "algorithm": "Ed25519",
    "original_message_verified": verified,
    "tampered_message_verified": tampered_verified,
    "tampered_message_rejected": not tampered_verified,
    "result": "PASS" if verified and not tampered_verified else "FAIL"
}

with open(
    os.path.join(RESULTS_DIR, "signature_verification.json"),
    "w",
    encoding="utf-8"
) as file:
    json.dump(signature_results, file, indent=4)

print("Signature verification results saved.")
```

#### Build the anomaly-detection system

```python
#Step 32 – Create the anomaly detector



from datetime import datetime, timezone, timedelta
from collections import defaultdict, deque
import json
import os

ANOMALY_LOG = os.path.join(
    RESULTS_DIR,
    "anomaly_alerts.json"
)

anomaly_alerts = []


class SecureNovaAnomalyDetector:

    def __init__(self):
        self.request_history = defaultdict(deque)
        self.previous_scope = {}
        self.used_tokens = set()
        self.volume_alert_active = set()

    def create_alert(self, identity, event_type, details, timestamp):

        alert = {
            "timestamp": timestamp.isoformat(),
            "identity": identity,
            "event_type": event_type,
            "details": details
        }

        anomaly_alerts.append(alert)

        print(
            f"ALERT | {alert['timestamp']} | "
            f"{identity} | {event_type}"
        )

    def process_request(
        self,
        identity,
        timestamp,
        scope,
        token_id=None,
        token_exp=None
    ):

        # Rule 1: More than 20 requests within 60 seconds
        history = self.request_history[identity]
        history.append(timestamp)

        while history and (
            timestamp - history[0]
        ).total_seconds() > 60:
            history.popleft()

        if len(history) > 20:

            if identity not in self.volume_alert_active:

                self.create_alert(
                    identity,
                    "REQUEST_VOLUME_SPIKE",
                    f"{len(history)} requests within 60 seconds",
                    timestamp
                )

                self.volume_alert_active.add(identity)

        else:
            self.volume_alert_active.discard(identity)

        # Rule 2: Detect scope changes
        if identity in self.previous_scope:

            old_scope = self.previous_scope[identity]

            if old_scope != scope:

                self.create_alert(
                    identity,
                    "SCOPE_CHANGE",
                    f"Scope changed from {old_scope} to {scope}",
                    timestamp
                )

        self.previous_scope[identity] = scope

        # Rule 3: Detect reuse of an expired token
        if token_id and token_exp:

            if token_id in self.used_tokens:

                if timestamp > token_exp:

                    self.create_alert(
                        identity,
                        "EXPIRED_TOKEN_REUSE",
                        f"Previously used token {token_id} presented after expiry",
                        timestamp
                    )

            self.used_tokens.add(token_id)


# Initialize detector
detector = SecureNovaAnomalyDetector()

print("SecureNova anomaly detector initialized successfully.")
print("Detection rules configured: 3")

```
