```python

import os
from openpyxl import Workbook
from openpyxl.styles import Font, PatternFill, Alignment, Border, Side
from openpyxl.utils import get_column_letter

# ----------------------------------------
# CREATE WORKBOOK
# ----------------------------------------

wb = Workbook()

ws1 = wb.active
ws1.title = "NIST AI RMF"

ws2 = wb.create_sheet("OWASP LLM Top 10")

# ----------------------------------------
# COMMON FORMATTING
# ----------------------------------------

dark_blue = "17365D"
light_blue = "D9EAF7"
light_green = "E2F0D9"
light_red = "FCE4D6"
white = "FFFFFF"

def format_sheet(ws, widths):

    ws.freeze_panes = "A4"
    ws.sheet_view.showGridLines = False

    for col, width in enumerate(widths, 1):
        ws.column_dimensions[
            get_column_letter(col)
        ].width = width

    for row in ws.iter_rows():

        for cell in row:

            cell.alignment = Alignment(
                vertical="top",
                wrap_text=True
            )

            cell.border = Border(
                bottom=Side(
                    style="thin",
                    color="D9E1F2"
                )
            )

    ws.sheet_properties.pageSetUpPr.fitToPage = True


def add_title(ws, title, subtitle, last_col):

    ws.merge_cells(
        start_row=1,
        start_column=1,
        end_row=1,
        end_column=last_col
    )

    ws.cell(1, 1, title)
    ws.cell(1, 1).font = Font(
        bold=True,
        size=16,
        color=white
    )
    ws.cell(1, 1).fill = PatternFill(
        "solid",
        fgColor=dark_blue
    )

    ws.merge_cells(
        start_row=2,
        start_column=1,
        end_row=2,
        end_column=last_col
    )

    ws.cell(2, 1, subtitle)
    ws.cell(2, 1).font = Font(
        italic=True,
        size=10,
        color="404040"
    )


def add_header(ws, row, headers):

    for col, value in enumerate(headers, 1):

        cell = ws.cell(row, col, value)

        cell.font = Font(
            bold=True,
            color=white
        )

        cell.fill = PatternFill(
            "solid",
            fgColor=dark_blue
        )

        cell.alignment = Alignment(
            horizontal="center",
            vertical="center",
            wrap_text=True
        )


# ========================================
# TAB 1: NIST AI RMF
# ========================================

add_title(
    ws1,
    "SECURENOVA AI - NIST AI RMF COMPLIANCE MAPPING",
    "Framework: NIST AI RMF 1.0 | Scope: Projects 1-4 | Self-assessment",
    6
)

nist_headers = [
    "Project",
    "Implemented Control / Activity",
    "NIST AI RMF Function",
    "Compliance Alignment",
    "Evidence Artifact",
    "Assessment"
]

add_header(ws1, 4, nist_headers)

nist_rows = [

    [
        "Project 1",
        "Six-identity inventory covering human user, AI agent, OAuth M2M client, LLM API key, RAG service and MCP server.",
        "GOVERN",
        "Establishes identity ownership, accountability and governance boundaries.",
        "Project 1 Identity Inventory; Threat Model",
        "Documented"
    ],

    [
        "Project 1",
        "Data Flow Diagram with three trust boundaries and six major data flows.",
        "MAP",
        "Identifies system context, actors, components and trust boundaries.",
        "Project 1 DFD; Trust Boundary Diagram",
        "Documented"
    ],

    [
        "Project 1",
        "STRIDE analysis covering 36 threat scenarios and risk-ranked threat register.",
        "MAP",
        "Identifies threats, vulnerabilities and potential impacts across system flows.",
        "Project 1 STRIDE Matrix; Risk Register",
        "Documented"
    ],

    [
        "Project 1",
        "Risk prioritization and ownership assignment.",
        "GOVERN",
        "Supports risk ownership and accountability.",
        "Project 1 Risk Register",
        "Documented"
    ],

    [
        "Project 2",
        "Auth0 identity management with OIDC, Authorization Code Flow and PKCE.",
        "MANAGE",
        "Applies authentication controls to reduce identity compromise risk.",
        "Project 2 Auth0 Configuration; Flask Authentication Evidence",
        "Implemented"
    ],

    [
        "Project 2",
        "RS256 JWT validation, audience and claim checks, MFA and role/scope restrictions.",
        "MANAGE",
        "Enforces identity verification and authorization controls.",
        "Project 2 JWT Evidence; MFA Configuration; 403 Test",
        "Implemented"
    ],

    [
        "Project 2",
        "M2M token expiry, refresh-token rotation and replay rejection tests.",
        "MEASURE",
        "Tests authentication lifecycle controls and detects token misuse.",
        "Project 2 M2M Token Test; Refresh Rotation Evidence",
        "Tested"
    ],

    [
        "Project 3",
        "Five red-team attack categories including indirect injection, identity spoofing, system prompt extraction, privilege escalation and RAG/MCP abuse.",
        "MAP",
        "Identifies adversarial AI attack pathways and their potential consequences.",
        "Project3_RedTeam.ipynb; attack_results/attack_log.jsonl; CVSS Findings",
        "Tested"
    ],

    [
        "Project 3",
        "CVSS scoring, MITRE ATLAS mapping and attack success matrix.",
        "MEASURE",
        "Assesses and prioritizes identified attack risks.",
        "cvss_findings.csv; mitre_atlas_mapping.json; attack_success_matrix.csv",
        "Documented"
    ],

    [
        "Project 4",
        "NeMo Guardrails tested against the five Project 3 attack scenarios.",
        "MANAGE",
        "Implements mitigation and tests protective responses to adversarial inputs.",
        "Project4_Before_After_Comparison.csv; Project4_Final_Results.json",
        "Tested"
    ],

    [
        "Project 4",
        "JWT sensitive-information redaction and Ed25519 message signing with tamper rejection.",
        "MANAGE",
        "Reduces credential disclosure and protects message integrity.",
        "Project 4 JWT Redaction Evidence; Ed25519 Verification Evidence",
        "Tested"
    ],

    [
        "Project 4",
        "Short-lived Auth0 access tokens, refresh-token rotation and risk-based authentication.",
        "MANAGE",
        "Reduces token exposure and unauthorized identity use.",
        "Project 4 Auth0 Configuration; Token Rotation Evidence",
        "Implemented"
    ],

    [
        "Project 4",
        "Anomaly detection, security alerts and five-attack replay comparison.",
        "MEASURE",
        "Evaluates detection effectiveness and records security events.",
        "project3_attack_replay_results.json; project3_attack_alerts.json; Project4_Final_Report.txt",
        "Tested"
    ]
]

for row in nist_rows:
    ws1.append(row)

# ----------------------------------------
# THREE REMAINING COMPLIANCE GAPS
# ----------------------------------------

gap_start = len(nist_rows) + 6

ws1.merge_cells(
    start_row=gap_start,
    start_column=1,
    end_row=gap_start,
    end_column=6
)

ws1.cell(
    gap_start,
    1,
    "REMAINING COMPLIANCE GAPS AND REMEDIATION PLAN"
)

ws1.cell(gap_start, 1).font = Font(
    bold=True,
    size=14,
    color=white
)

ws1.cell(gap_start, 1).fill = PatternFill(
    "solid",
    fgColor="9C0006"
)

gap_headers = [
    "Finding",
    "Obligation / Expected Practice",
    "Risk",
    "Recommendation",
    "Priority",
    "Status"
]

add_header(ws1, gap_start + 1, gap_headers)

gaps = [

    [
        "GAP-01: No demonstrated automated periodic agent identity recertification.",
        "Maintain an accurate identity inventory and periodically confirm that every agent still requires its permissions.",
        "Dormant or overprivileged agents may retain unauthorized access.",
        "Implement automated 30-day privileged and 90-day standard identity reviews, with owner approval and audit records.",
        "HIGH",
        "Open"
    ],

    [
        "GAP-02: No demonstrated centralized secrets-manager audit and automated key-rotation enforcement.",
        "Store credentials securely, restrict access and enforce API key rotation within 90 days.",
        "Exposed or stale API credentials may enable unauthorized LLM or service access.",
        "Deploy a managed secrets vault, secret scanning, automated rotation and access auditing.",
        "HIGH",
        "Open"
    ],

    [
        "GAP-03: No demonstrated independent RAG source-integrity validation and end-to-end data-exfiltration monitoring.",
        "Validate retrieved information and monitor sensitive-data access and outbound transfers.",
        "Poisoned knowledge sources or manipulated agent workflows may cause misinformation or data disclosure.",
        "Add trusted-source validation, retrieval provenance, output verification, egress monitoring and recurring incident exercises.",
        "HIGH",
        "Open"
    ]
]

for row in gaps:
    ws1.append(row)

# ----------------------------------------
# TAB 2: OWASP LLM TOP 10 (2025)
# ----------------------------------------

add_title(
    ws2,
    "SECURENOVA AI - OWASP LLM TOP 10 COMPLIANCE MAPPING",
    "Reference: OWASP Top 10 for LLM Applications 2025",
    6
)

owasp_headers = [
    "OWASP Category",
    "Risk Name",
    "Project 3 Attack",
    "Project 4 Defensive Control",
    "Evidence Artifact",
    "Coverage Assessment"
]

add_header(ws2, 4, owasp_headers)

owasp_rows = [

    [
        "LLM01:2025",
        "Prompt Injection",
        "Indirect Prompt Injection using hidden-document payloads attempting to disclose a simulated JWT.",
        "NeMo Guardrails blocks the tested attack; JWT redaction reduces sensitive-token exposure; security alerts record blocked attempts.",
        "Project3 attack_log.jsonl; project3_attack_replay_results.json; Project4_Before_After_Comparison.csv",
        "Tested - five-attack replay"
    ],

    [
        "LLM06:2025",
        "Excessive Agency",
        "MCP abuse and prompt-based privilege escalation attempting unauthorized agent actions.",
        "NeMo Guardrails, Auth0 scope restrictions, short-lived tokens, refresh-token rotation and anomaly detection.",
        "Project 3 MCP Attack Evidence; Project 2 Scope/403 Evidence; Project4_Final_Results.json",
        "Partial - control tests available; broader tool authorization review needed"
    ],

    [
        "LLM07:2025",
        "System Prompt Leakage",
        "System prompt extraction attempts using instruction override, role-play and related techniques.",
        "NeMo Guardrails blocks tested extraction attempts; JWT redaction protects credential-like sensitive content.",
        "Project 3 System Prompt Extraction Evidence; Project4_Before_After_Comparison.csv",
        "Tested - known attack cases; broader prompt confidentiality review needed"
    ],

    [
        "LLM09:2025",
        "Misinformation",
        "RAG poisoning attempt intended to manipulate retrieved knowledge and produce misleading responses.",
        "NeMo Guardrails provides a mitigation layer; independent source verification and factuality validation remain a gap.",
        "Project 3 RAG Poisoning Evidence; Project4_Final_Report.txt",
        "Partial - source integrity and factuality controls need strengthening"
    ]
]

for row in owasp_rows:
    ws2.append(row)

# Additional identity threat clarification

note_row = len(owasp_rows) + 6

ws2.merge_cells(
    start_row=note_row,
    start_column=1,
    end_row=note_row,
    end_column=6
)

ws2.cell(
    note_row,
    1,
    "ADDITIONAL IDENTITY THREAT CROSS-REFERENCE"
)

ws2.cell(note_row, 1).font = Font(
    bold=True,
    color=white
)

ws2.cell(note_row, 1).fill = PatternFill(
    "solid",
    fgColor=dark_blue
)

ws2.merge_cells(
    start_row=note_row + 1,
    start_column=1,
    end_row=note_row + 2,
    end_column=6
)

ws2.cell(
    note_row + 1,
    1,
    "Project 3 Agent Identity Spoofing is an identity and authentication threat. "
    "It is cross-referenced to Project 2 Auth0 identity validation, RS256 JWT "
    "validation and authorization controls, and Project 4 token rotation and "
    "anomaly detection. It is not labelled as OWASP LLM09:2025, which means Misinformation."
)

ws2.cell(note_row + 1, 1).alignment = Alignment(
    wrap_text=True,
    vertical="center"
)

# ----------------------------------------
# FORMAT BOTH WORKSHEETS
# ----------------------------------------

format_sheet(
    ws1,
    [19, 48, 19, 48, 48, 18]
)

format_sheet(
    ws2,
    [18, 25, 48, 52, 48, 35]
)

ws1.auto_filter.ref = f"A4:F{4 + len(nist_rows)}"
ws2.auto_filter.ref = f"A4:F{4 + len(owasp_rows)}"

ws1.row_dimensions[1].height = 30
ws2.row_dimensions[1].height = 30

for ws in [ws1, ws2]:

    for row in ws.iter_rows(min_row=5):

        for cell in row:

            if cell.row % 2 == 0:
                cell.fill = PatternFill(
                    "solid",
                    fgColor="F2F6FA"
                )

# Highlight the three gaps
for row_num in range(gap_start + 2, gap_start + 5):

    for cell in ws1[row_num]:

        cell.fill = PatternFill(
            "solid",
            fgColor=light_red
        )

# Highlight assessment cells
for row_num in range(5, 5 + len(nist_rows)):

    ws1.cell(row_num, 6).fill = PatternFill(
        "solid",
        fgColor=light_green
    )

# ----------------------------------------
# SAVE WORKBOOK
# ----------------------------------------

output_file = os.path.join(
    compliance_path,
    "SecureNova_AI_Compliance_Mapping.xlsx"
)

wb.save(output_file)

print("======================================")
print("COMPLIANCE WORKBOOK CREATED")
print("======================================")
print("File:", os.path.abspath(output_file))
print("Worksheets:", wb.sheetnames)
print("NIST mappings:", len(nist_rows))
print("OWASP mappings:", len(owasp_rows))
print("Remaining compliance gaps:", len(gaps))
print("======================================")
```
